## Lab 1 - Bài 4

```python
# code lab 1 bài 4 ở đây
import math


a = float(input("Nhập a: "))
b = float(input("Nhập b: "))
c = float(input("Nhập c: "))


delta = b * b - 4 * a * c


print("Delta =", delta)

if delta >= 0:
    can_delta = math.sqrt(delta)
    print("Căn delta =", can_delta)
else:
    print("Delta < 0 nên không có căn bậc hai thực")
```

## Lab 1 - Bài 5: bài tập làm thêm

```python
# code lab 1  bài làm thêm ở đây 
import math

print("1.Cộng/Trừ/Nhân/Chia\n2.Lũy thừa\n3.Căn bậc hai\n4.Hàm lượng giác\n5.Logarit")
ch = int(input("Chọn số: "))

if ch == 1:
    a, b = map(float, input("Nhập a b: ").split())
    print("Cộng:", a+b, "Trừ:", a-b, "Nhân:", a*b, "Chia:", a/b)

elif ch == 2:
    x, y = map(float, input("Nhập x y: ").split())
    print("x^y =", x**y)

elif ch == 3:
    x = float(input("Nhập x: "))
    print("√x =", math.sqrt(x))

elif ch == 4:
    deg = float(input("Nhập góc (độ): "))
    r = math.radians(deg)
    print("sin:", math.sin(r), "cos:", math.cos(r), "tan:", math.tan(r))

elif ch == 5:
    x = float(input("Nhập x: "))
    print("log10:", math.log10(x), "ln:", math.log(x))
    base = float(input("Nhập cơ số: "))
    print(f"log cơ số {base}:", math.log(x, base))


```
